# Error Management: Gestione degli errori

### Errore non gestito

```python
print("INIZIO")
my_value = int("CIAO")
print("Continua..")
```

Come possiamo notare, il programma stampa INIZIO, poi va in errore.
L’errore è un ValueError: Python infatti non può convertire la stringa "CIAO" in un numero intero e quindi interrompe l’esecuzione. Di conseguenza, la riga print("Continua..") non viene mai eseguita.

#### Gestiamo la nostra eccezione
A questo punto vediamo una soluzione minimale per gestire gli errori usando le keyword try ed except.
Dentro il blocco try inseriamo il codice che vogliamo eseguire.
Dentro il blocco except invece indichiamo cosa fare in caso di errore.
Quando eseguiamo il codice, se tutto funziona correttamente, viene eseguito il contenuto del try. Se invece si verifica un errore, Python interrompe l’esecuzione del try ed entra nell’except, senza far terminare l’applicazione. In entrambi i casi, il programma continua a funzionare dopo la gestione dell’errore.

```python
try:
    print("INIZIO")
    my_value = int("CIAO")
    print("FINE TRY")
except:
    print("Si e' verificato un errore")

print("Continua..")
```

```python
try:
    print("INIZIO")
    my_value = int("1")
    print("FINE TRY")
except:
    print("Si e' verificato un errore")

print("Continua..")
```

#### A questo punto introduciamo una nuova keyword: finally

Il blocco finally viene eseguito sempre, indipendentemente dal fatto che ci sia stato un errore o meno.
Questo è utile, ad esempio, quando vogliamo rilasciare delle risorse, chiudere file o connessioni, o semplicemente eseguire del codice finale garantito.

```python
try:
    print("INIZIO")
    my_value = int("CIAO")
    print("FINE TRY")
except:
    print("Si e' verificato un errore")
finally:
    print("Eseguito sempre dopo il try/except")
    
print("Continua..")
```

```python
try:
    print("INIZIO")
    my_value = int("1")
    print("FINE TRY")
except:
    print("Si e' verificato un errore")
finally:
    print("Eseguito sempre dopo il try/except")
    
print("Continua..")
```

#### Vediamo l’ultima keyword collegata al'error management: else

L’else è simile al finally, ma con una differenza importante: il suo contenuto viene eseguito solo se non si sono verificati errori.
Personalmente non sono un grande fan di questa keyword, perché non tutti i linguaggi ce l’hanno (ad esempio in JavaScript non esiste), ma è comunque uno strumento che Python ci mette a disposizione.

```python
try:
    print("INIZIO")
    my_value = int("CIAO")
    print("FINE TRY")
except:
    print("Si è verificato un errore")
else:
    print("Codice eseguito solo se non ci sono errori")
finally:
    print("Eseguito sempre dopo il try/except")

print("Continua..")
```

#### Gestire errori diversi in maniera differente

Fino ad ora abbiamo visto come gestire il ValueError, ma soprattutto abbiamo visto che con except possiamo catturare qualsiasi tipo di errore.
Nella realtà però, durante il ciclo di vita di uno script o di un’applicazione, possono verificarsi molti errori diversi.
Nella documentazione ufficiale di Python (vi lascio il link in descrizione) trovate una lista dei principali errori di base. Altri tipi di errori si aggiungono con le librerie: ad esempio, la libreria requests che abbiamo usato in un video precedente espone errori specifici legati alla comunicazione con le API.

Ora aggiungiamo un altro errore molto comune: lo ZeroDivisionError. Se provo a fare 10/0, Python mi segnala che non è possibile dividere per zero.
Vediamo quindi come catturare sia il ValueError che lo ZeroDivisionError nel nostro programma.

```python
10/0
```

In questo codice chiediamo all'utente l'input di my_value:
- Se scriviamo "Ciao" => Causiamo un ValueError
- Se scriviamo "0" => Il numero viene convertito in intero, ma causiamo un ZeroDivisionError
- Se scriviamo 10 => Il try viene eseguito con successo

```python
try:
    print("INIZIO")
    my_value = int(input("Inserisci un numero: "))
    result = 10 / my_value
    print(f"Risultato operazione: {result}")
except:
    print("Si è verificato un errore")
else:
    print("Codice eseguito solo se non ci sono errori")
finally:
    print("Eseguito sempre dopo il try/except")

print("Continua..")
```

Funziona benissimo, tuttavia possiamo notare che viene eseguito lo stesso except sia in caso di ValueError che in caso di ZeroDivisionError. Non e' ideale, sopratutto quando si prova ad analizzare i log per capire il problema.

Possiamo quindi utilizzare except dedicati per ogni errore e mantenere l'except attuale come errore generico:

```python
try:
    print("INIZIO")
    my_value = int(input("Inserisci un numero: "))
    result = 10 / my_value
    print(f"Risultato operazione: {result}")
    # print(gigi)
    
except ValueError:
    print("Errore: non hai inserito un numero valido.")
    
except ZeroDivisionError:
    print("Errore: non puoi dividere per zero.")

except NameError:
    print("Errore: stai stampando una variabile inesistente")
    
except:
    print("Errore generico")
    
else:
    print("Codice eseguito solo se non ci sono errori")
    
finally:
    print("Eseguito sempre dopo il try/except")


print("Continua..")
```
