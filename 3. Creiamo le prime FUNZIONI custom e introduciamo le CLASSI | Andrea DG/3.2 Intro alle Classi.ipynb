# 🧰 Classi
#### Una classe è una specie di stampino, per creare oggetti più complessi. Ci permette di dire:
#### “Ehi Python, esiste un tipo di dato chiamato Bevanda, ed ecco come funziona”

### 1️⃣ Definiamo una classe utilizzando la keyword **class** e definiamo il costruttore
Il costruttore e' un metodo speciale che viene eseguito automaticamente quando si crea un'istanza di una classe. Serve per inizializzare gli attributi dell'oggetto con valori specifici al momento della creazione.

```python
class Bevanda:
    # costruttore
    def __init__(self, nome, litri, zucchero):
        self.nome = nome
        self.litri = litri
        self.zucchero = zucchero
```

### 2️⃣ Instanziamo qualche oggetto di tipo Bevanda

```python
succo_arancia = Bevanda(nome="Succo d'arancia bio", litri=0.5, zucchero=10)
```

### 3️⃣ Definiamo il metodo di istanza "descrivi"
Un metodo di istanza è una funzione definita all'interno di una classe che opera sui dati dell’istanza (oggetto) e ha accesso a essi tramite il primo parametro, di solito chiamato self.

```python
class Bevanda:
    # costruttore
    def __init__(self, nome, litri, zucchero):
        self.nome = nome
        self.litri = litri
        self.zucchero = zucchero

    # creato un metodo di istanza (utilizza self)
    def descrivi(self):
        print(f"Nome Bevanda: {self.nome} - Litri: {self.litri} - Zucchero: {self.zucchero}gr")

succo_arancia = Bevanda(nome="Succo d'arancia bio", litri=0.5, zucchero=10)
succo_arancia.descrivi()
```

### 4️⃣ Definiamo due metodi di classe
Un metodo di classe è una funzione definita in una classe che agisce sulla classe stessa e non sulle singole istanze; riceve la classe come primo argomento, convenzionalmente chiamato cls, ed è decorato con @classmethod.

```python
class Bevanda:
    # costruttore
    def __init__(self, nome, litri, zucchero):
        self.nome = nome
        self.litri = litri
        self.zucchero = zucchero

    # creato un metodo di istanza (utilizza self)
    def descrivi(self):
        print(f"Nome Bevanda: {self.nome} - Litri: {self.litri} - Zucchero: {self.zucchero}gr")

    # metodo di classe (utilizza cls)
    @classmethod
    def crea_coca_cola_zero(cls):
        return cls(nome="Coca Cola zero", litri=0.33, zucchero=0)

    # metodo di classe (utilizza cls)
    @classmethod
    def crea_fanta(cls):
        return cls(nome="Fanta", litri=0.33, zucchero=0)

# Istanziato un oggetto usando il metodo di classe crea_coca_cola_zero
coca_zero_1 = Bevanda.crea_coca_cola_zero()
coca_zero_2 = Bevanda.crea_coca_cola_zero()

# Istanziato un oggetto usando il metodo di classe crea_fanta
fanta_1 = Bevanda.crea_fanta()
fanta_2 = Bevanda.crea_fanta()

# Istanziato un oggetto usando il costruttore
succo_strano = Bevanda(nome="Succo d'arancia bio", litri=0.5, zucchero=10)

# Metodo di istanza descrivi richiamato su oggetto succo_strano
succo_strano.descrivi()
```
