# ** Cicli e Loop Iterations**
- il ciclo `while`, che esegue un blocco finché una condizione è vera;
- il ciclo `for`, che ci permette di iterare su una sequenza;
- i comandi `break` e `continue` per controllare il flusso;
- le **list comprehension**, un modo elegante per creare nuove liste;
- e la funzione `input`, che useremo per rendere il nostro programma interattivo.

#### **While** per popolare una lista di numeri

```python
lista_numerica = []
contatore = 1

while contatore <= 10:
    print(f"Aggiungi valore {contatore} a lista_numerica")
    lista_numerica.append(contatore)
    contatore = contatore + 1
    
print(f"Risultato: {lista_numerica}")
```

#### Un esempio concreto

```python
lista_numeri_pari = []
lista_numeri_dispari = []
contatore = 1

while contatore <= 20:
    if(contatore % 2 == 0):
        lista_numeri_pari.append(contatore)
    else:
        lista_numeri_dispari.append(contatore)
    contatore += 1

print(f"Lista numeri pari: {lista_numeri_pari}")
print(f"Lista numeri dispari: {lista_numeri_dispari}")
```

#### Usiamo while per calcolare il doppio di tutti i valori in un range

```python
lista = [1, 2]
len(lista)
```

```python
range_valori = range(1, 10 + 1)
lista_valori_doppi = []
contatore = 0

while contatore < len(range_valori):
    doppio = range_valori[contatore] * 2
    lista_valori_doppi.append(doppio)
    contatore += 1

print(list(range_valori))
print(lista_valori_doppi)
```

#### While con input

```python
input("Inserisci utente (scrivi \"esci\" per uscire): ")
```

```python
lista_utenti = []
comando = ""

while True:
    comando = input("Inserisci utente (scrivi \"esci\" per uscire): ")
    if(comando == "esci"):
        break
    lista_utenti.append(comando)
        
print(lista_utenti)
```

#### For loop

```python
lista_bevande = ["Cola", "Aranciata", "Succo di mele"]

for bevanda in lista_bevande:
    if bevanda == "Aranciata":
        continue
    print(f"Bevanda: {bevanda}")

print("Fine loop")
```

#### For Loop utilizzando il range

```python
numeri_dispari = []
for numero in range(1, 11):
    if numero % 2 != 0:
        numeri_dispari.append(numero)

print(numeri_dispari)
```

#### List Comprehension semplice

```python
doppio = [numero * 2 for numero in range(1, 11)]
print(doppio)
```

#### Con if

```python
doppio_fino_a_dieci = [numero * 2 for numero in range(1, 11) if numero <= 5]
print(doppio_fino_a_dieci)
```

#### Con if else

```python
acque = [
    {"nome": "Acqua Brand 1", "prezzo": 1.22},
    {"nome": "Acqua Brand 2", "prezzo": 1.43},
    {"nome": "Acqua Brand 3", "prezzo": 3},
    {"nome": "Acqua Brand 4", "prezzo": 2.01},
    {"nome": "Acqua Brand 5", "prezzo": 1.99},
    {"nome": "Acqua Brand 6", "prezzo": 1.76}
]

etichetta_prezzi = [
    {"nome": acqua["nome"], "etichetta_prezzo": "Cara"} if acqua["prezzo"] > 1.5
    else {"nome": acqua["nome"], "etichetta_prezzo": "Ok"} 
    for acqua in acque
]

for acqua_etichettata in etichetta_prezzi:
    print(f"{acqua_etichettata["nome"]} e' {acqua_etichettata["etichetta_prezzo"]}")
```

#### Refactoring 

```python
acque = [
    {'nome': 'Acqua Brand 1', 'prezzo': 1.22},
    {'nome': 'Acqua Brand 2', 'prezzo': 1.43},
    {'nome': 'Acqua Brand 3', 'prezzo': 3.00},
    {'nome': 'Acqua Brand 4', 'prezzo': 2.01},
    {'nome': 'Acqua Brand 5', 'prezzo': 1.99},
    {'nome': 'Acqua Brand 6', 'prezzo': 1.76}
]

etichettate = [
    {
        'nome': acqua['nome'],
        'etichetta': 'Cara' if acqua['prezzo'] > 2 else 'Ok'
    }
    for acqua in acque
]

for item in etichettate:
    print(f"{item['nome']} è {item['etichetta']}")
```

#### Esempio finale

```python
lista_spesa = [
    {
        "nome": "Pane",
        "prezzo": 2.00,
        "quantita": 1
    },
    {
        "nome": "Acqua",
        "prezzo": 1,
        "quantita": 6
    },
    {
        "nome": "Gelato",
        "prezzo": 7,
        "quantita": 1
    },
]

def calcola_spesa(lista_spesa):
    """
    Calcola l'ammontare della spesa considerando la quantita acquistata e il prezzo del singolo elemento
    """
    totale = 0
    for oggetto_acquistato in lista_spesa:
        totale += (oggetto_acquistato["prezzo"] * oggetto_acquistato["quantita"])

    return totale

def calcola_sconto(totale, sconto):
    """
    Ritorna il valore scontato e l'ammontare dello sconto
    """
    totale_scontato = totale - (totale * sconto) / 100
    ammontare_sconto = totale - totale_scontato
    return {
        "totale_scontato": totale_scontato,
        "ammontare_sconto": ammontare_sconto
    }

totale = calcola_spesa(lista_spesa)
totale_scontato = calcola_sconto(totale, 10)

print(f"Totale: {totale}")
print(f"Con sconto di -{totale_scontato["ammontare_sconto"]} paghi: {totale_scontato["totale_scontato"]}")
```
