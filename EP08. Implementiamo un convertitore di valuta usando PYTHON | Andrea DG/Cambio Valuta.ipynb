### Scriviamo il nostro programmino
Let's have fun!

```python
import requests

response = requests.get("https://api.frankfurter.dev/v1/latest?base=USD")
data = response.json()
rates = data["rates"]
base = data["base"]

key_value_changes = [f"{base}/{rate}" for rate in rates.keys()]

print("Seleziona il cambio valuta desiderato:")
[print(f"{index}. {coppia}") for index, coppia in enumerate(key_value_changes, start=1)]
print("Scrivi \"esci\" per terminare il programma")

while True:
    requested_change = input("Inserisci il valore desiderato: ")
    if(requested_change == "esci"):
        break
    selected_change = key_value_changes[int(requested_change)-1]
    valore_di_cambio = selected_change.split("/")[1]
    print(f"Il valore di cambio per {selected_change} e' di {rates[valore_di_cambio]}")
```
