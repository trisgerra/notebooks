# Collezioni di dati in Python
Oggi parliamo delle principali collezioni in Python:

### Liste
- I dati sono contenuti tra [];
- Sono ordinate: la posizione dei dati contenuti non cambia se non viene modificata;
- Indicizzate: Ogni elemento e' assegnato ad un indice
- Mutabile: Possiamo modificarne il contenuto aggiungendo, cancellando e aggiornando i valori
- Permette duplicati: possiamo avere dati duplicati all'interno della lista

```python
la_mia_lista = [1, 2, 3, 3, "Ciao", True, 2.2]

print(la_mia_lista)                                         # Ritorna [1, 2, 3, 3, "Ciao", True, 2.2]
type(la_mia_lista)                                          # Ritorna list
print(la_mia_lista[5])                                      # Ritorna il dato all'indice 5, True

# Modifichiamo la lista
print(f"La mia lista prima della modifica: {la_mia_lista}") # Stampa la variabile_la_mia_lista
la_mia_lista[0] = "Buonanotte"                              # Riscrive il valore all'indice 0
print(f"La mia lista dopo la modifica: {la_mia_lista}")     # Stampa la lista aggiornata

valore_indice_0 = la_mia_lista.pop(0)                       # Rimuoviamo il valore all'indice 0 e lo assegniamo a valore_indice_o

print(valore_indice_0)                                      # Stampiamo la variabile contenente il valore rimosso
print(la_mia_lista)                                         # Stampiamo la lista dopo aver rimosso il valore

# Vediamo la documentazione per il metodo insert
la_mia_lista.insert?
```

```python
lista_numerica = [1,2,4]      # Creiamo una lista numerica
lista_numerica.insert(2, 3)   # Usiamo insert per inserire un valore all'indice 2 spostando il valore 4 all'indice 3
print(lista_numerica)       
```

### Tuple
- I dati sono contenuti tra ();
- Sono ordinate: la posizione dei dati contenuti non cambia se non viene modificata;
- Indicizzate: Ogni elemento e' assegnato ad un indice
- NON Mutabile: Non possiamo modificarne il contenuto aggiungendo, cancellando e aggiornando i valori dopo averla istanziata
- Permette duplicati: possiamo avere dati duplicati all'interno della lista

```python
la_mia_tupla = (0, 2, 3, 3, "Ciao", 2.2) # Istanziamo una tupla 
la_mia_tupla.count("Ciao")               # Verifichiamo quante volte contiene il valore "Ciao"

# Utilizziamo *args della nostra funzione, *args e' una tupla
def la_mia_funzione(*args):
    # args[0] = "Giorgio"                # Se togliamo il commento da questa riga, fallisce, non possiamo modificare una tupla 
    print(args)

la_mia_funzione("Andrea", "Ciccio")      # Chiamiamo la funzione passandogli i notri argomenti
```

### Set
- I dati sono contenuti tra {};
- NON Sono ordinate: la posizione dei dati non e' garantita;
- NON Sono Indicizzate: Non possiamo accedere tramite l'indice
- Mutabile: Possiamo modificarne il contenuto nonostante non possiamo utilizzare l'indice
- NON Permette duplicati: possiamo passare dei duplicati ma verranno rimossi dal Set

```python
# Creiamo una LISTA di bevande simile a un distributore automatico
lista_bevande = ["CocaCola", "CocaCola", "CocaCola", "CocaCola", "Fanta","Fanta","Fanta", "Sprite", "CocaCola"]

# Stampiamo la lista e verifichiamo che contiene piu' volte la stessa bevanda
print(lista_bevande)

# Castiamo la lista in set, ovvero trasformiamolo da Lista a Set
set_bevande = set(lista_bevande)

# Stampiamo il nostro Set, e notiamo che adesso tutti i duplicati vengono rimossi
print(set_bevande)

# Mutiamo il set aggiungendo una nuova bevanda e lo stampiamo
set_bevande.add("Spremuta d'arancia")
print(set_bevande)

# Rimuoviamo la Fanta dal set e stampiamo per verificarne la rimozione
set_bevande.discard("Fanta")
print(set_bevande)

# Utilizziamo il metodo Union per unire 2 nuovi set al nostro set_bevande e creiamo un nuovo set dal risultato
nuovo_set = set_bevande.union({"Cappuccino"}, {"Acqua", "Caffe"})
print(nuovo_set)
```

```python
# Vediamo la documentazione per  il metodo difference
set_bevande.difference?
```

```python
# Proviamo il metodo difference che ci ritorna tutti gli elementi
# Contenuti nel set dal quale chiamiamo il metodo (macchinetta_2) e non in quello passato come argomento (macchinetta_1)
macchinetta_1 = {"CocaCola", "Fanta", "Sprite"}
macchinetta_2 = {"CocaCola", "Fanta", "Caffe"}

print(macchinetta_2.difference(macchinetta_1))
```

### Range
- Viene creato con la funzione range();
- Sono ordinati: la posizione dei dati contenuti non cambia se non viene modificata;
- Indicizzato: Possiamo accedere all'elemento a un determinato indice in lettura
- Mutabile: Non possiamo mutarlo, ma possiamo Castarlo in lista e a quel punto possiamo anche mutarlo
- Permette duplicati: possiamo avere dati duplicati all'interno della lista

```python
mio_range = range(10)           # Creiamo un range da 0 a 10
print(list(mio_range))          # Facciamo il cast in lista e vediamo il contenuto
print(list(range(10, 21, 2)))   # Utilizziamo i parametri di configurazione per customizzarlo

# Stampiamo la documentazione per la funzione range
range?
```

```python
# mio_range[2] = "Fallisce"        # Togliamo il commento qui per vedere come i range non sono mutabili, fallira'

lista_da_range = list(mio_range)   # Convertiamo il range in lista
lista_da_range[2] = 10             # Riassegnamo il valore all'indice 2 della nostra lista
print(lista_da_range)              # Stampiamo la lista aggiornata con il valore 10 all'indice 2
```

### Dict
- I dati sono contenuti tra {} ma sono coppie chiave-valore;
- Sono ordinati nelle nuove versioni di Python: la posizione dei dati contenuti non cambia se non viene modificata;
- Indicizzati: Possiamo accedere all'elemento attraverso la Chiave;
- Mutabile: Possiamo aggiungere, modificare o rimuovere coppie Chiave-Valore;
- Non permette Chiavi duplicate: I valori possono essere duplicati, ma le chiavi devono essere univoche

```python
# Creiamo un dizionario
dizionario = {
    "nome": "Andrea",
    "cognome": "Rossi",
    "indirizzo": {
        "via": "Via Roma",
        "cap": 97199,
        "provincia": "LM" 
    }
}

# Stampiamo il valore della chiave nome
print(dizionario["nome"])

# Stampiamo il valore della chiave cap all'interno del sub-dizionario indirizzo
print(dizionario["indirizzo"]["cap"])

# Aggiungiamo una nuova coppia di chiave-valore
dizionario["nuovo"] = True

# Stampiamo il valore di nuovo, appena aggiunto
print(dizionario["nuovo"])

# Utilizziamo il metodo get per accedere a una variabile non esiste e ritorniamo il valore di Fallback, "Non Trovato"
dizionario.get("numero_di_telefono", "Non trovato")

# Stampiamo tutte le chiavi
print(dizionario.keys())

# Stampiamo tutti i valori
print(dizionario.values())
```
