## Operatori di comparazione (comparison operators)

- `>` maggiore  
- `<` minore  
- `<=` minore o uguale  
- `>=` maggiore o uguale  
- `==` uguale (**attenzione al doppio `=`, uno solo viene usato per assegnare il valore**)  
- `!=` diverso  
- `is` Identico (lo stesso oggetto in memoria)  
- `is not` Non Identico  
- `in` (es. con le liste): contiene  
- `not in` (es. con le liste): non contiene

```python
x = 5
print(x >= 5)
```

## Controllo di flusso

```python
x = -100

if x < 0:
    print("X e' minore di 0")
elif x < 10:
    print("X e' minore di 10")
elif x < 100:
    print("X e' maggiore di 10 e minore di 100")
else:
    print("X e' maggiore di 100")
```

```python
comando = "Ciccio"

match comando:
    case "Avvia Musica":
        print("Avvio")
    case "Ferma Musica":
        print("Stop")
    case _:
        print("Comando non riconosciuto")
```
