```python
def maggiorenne(eta):
    if(eta > 18):
        print("Maggiorenne")
    elif(eta == 18):
        print("Ha proprio 18 anni")
    else:
        print("Minorenne")
        
    return eta >= 18

maggiorenne(18)
```

### Operatori logici (and or not)

```python
def maggiorenne(eta, sesso = "maschile"):
    maggiorenne = eta >= 18
    giovane = eta < 25
    # adolescente = eta > 14 and eta < 18
    adolescente = 14 < eta < 18 # piu pulito e leggibile
    bambino = eta <= 14 or eta == 0
    is_donna = sesso.lower() == "femminile"

    if maggiorenne and giovane and is_donna:
        print("Ragazza")
    elif maggiorenne and giovane and not(is_donna):
        print("Ragazzo")
    elif eta == 18:
        print("Ha proprio 18 anni")
    elif adolescente:
        print("Adolescente")
    elif bambino:
        print("Minorenne")
    else:
        print("Età non gestita")

    return maggiorenne

maggiorenne(22)
```

### Nelle classi

```python
class Bevanda:
    # costruttore
    def __init__(self, nome, litri, zucchero, alcool):
        self.nome = nome
        self.litri = litri
        self.zucchero = zucchero
        self.alcool = alcool

    @classmethod
    def crea_fanta(cls):
        return cls(nome="Fanta", litri=0.33, zucchero=0, alcool=0)
    
    def descrivi(self):
        print(f"""
        Nome Bevanda: {self.nome}
        Litri: {self.litri} 
        Zucchero: {self.zucchero}gr
        Alcool: {self.alcool * 100}%""")

    def vendibile_ai_minorenni(self):
        if self.alcool == 0:
            return True
        else:
            print("Richiedi documento")
            # richiedi_documento()
            return False

fanta = Bevanda.crea_fanta()
succo = Bevanda("Succo d'arancia", 0.33, 5, 0.0)
birra = Bevanda("Andrew Beer", 0.33, 1, 0.06)
```

```python
birra.vendibile_ai_minorenni()
```
