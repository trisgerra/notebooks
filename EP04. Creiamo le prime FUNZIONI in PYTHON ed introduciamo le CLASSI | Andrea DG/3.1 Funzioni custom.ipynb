# 💡 Funzioni
#### Un blocco di codice che può essere eseguito multiple volte, permette di organizzare la logica del nostro applicativo, evitando duplicazioni.

### 1️⃣ Funzione semplice, senza parametri e senza return 
Esegue il codice definito all'interno dello scope (indentazione corretta).
Questa e' una funzione senza return (void), se assegnamo il valore ritornato dalla funzione a una variabile sara' None.

```python
def saluta():
    print("Ciao")

saluta()
```

```python
risultato_saluta = saluta()
print(f"Il valore di risultato_prova e': {risultato_saluta}")
```

### 2️⃣ Ritorniamo un valore (return)
In questo caso ritorniamo un valore utilizzando la keyword **return** , possiamo quindi assegnare il valore ritornato dalla funzione a una variabile.

```python
def ritorna_eta():
    # Immagina una riga di codice
    # Un altra riga
    return 18

print(ritorna_eta())
```

### 3️⃣ Funzione con un parametro
Possiamo definire la nostra funzione in modo che accetti un parametro in ingresso (nome).
In fase di utilizzo, quando richiamiamo la funzione, possiamo passarle l'argomento desiderato ("Mario", "Barbara").

```python
def saluta_nome(nome):
    print(f"Ciao {nome}")

saluta_nome("Mario")
saluta_nome("Barbara")
```

### 4️⃣ Funzione con due parametri
Possiamo passare diversi parametri. 
Inoltre possiamo definire il valore di default in fase di definizione della funzione, basta aggiungere ="Valore Di Default Desiderato"), ad esempio **nome** e **cognome** nell'esempio sotto hanno come valore di default "Mario" e "Rossi".

```python
def saluta_persona(nome="Mario", cognome="Rossi"):
    print(f"Ciao {nome} {cognome}")

saluta_persona("Verdi", "Andrea",)
saluta_persona(cognome="Bianchi", nome="Mario")
```

### 5️⃣ Passare un numero variabil di argomenti con *args
Come visto nella funzione join nel video precedente, possiamo fare in modo che la nostra funzione sia in grado di ricevere un numero variabile di argomenti, questi saranno gli *args. 
Questo parametro sara' una tupla nell'esempio sottostante e possiamo accedere all'argomento specifico utilizzando l'indice.

```python
def proviamo_args(*args):
    print(args[1])

proviamo_args("Andrea", "Gigi", "Maria")
```

### 6️⃣Passare un dizionario variabile con **kwargs
Possiamo anche utilizzare **kwargs per passare un dizionario con coppie chiave-valore variabili.

```python
def presentati(**kwargs):
    print(f"Ciao io sono {kwargs["nome"]} {kwargs["cognome"]} ed ho {kwargs["eta"]} anni.")

presentati(nome="Gigi", cognome="Rossi", eta=44)
```
